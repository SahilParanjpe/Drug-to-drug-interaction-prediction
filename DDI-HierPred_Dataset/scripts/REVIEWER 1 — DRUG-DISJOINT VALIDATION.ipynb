```python
from pathlib import Path
import pandas as pd

ROOT = Path(r"C:\Users\user\Desktop\drug drug\drug2")
SPLIT_DIR = ROOT / "results_level2_pair_disjoint" / "splits"

train_file = SPLIT_DIR / "Level2_Train_Pair_Disjoint.csv"
test_file  = SPLIT_DIR / "Level2_Test_Pair_Disjoint.csv"

print("=" * 70)
print("PAIR-DISJOINT SPLIT — DRUG IDENTITY OVERLAP AUDIT")
print("=" * 70)

print("\nTrain file:", train_file)
print("Exists:", train_file.exists())

print("\nTest file:", test_file)
print("Exists:", test_file.exists())

train = pd.read_csv(train_file)
test  = pd.read_csv(test_file)

print("\nTrain shape:", train.shape)
print("Test shape :", test.shape)

print("\nTrain columns:")
print(train.columns.tolist())

print("\nTest columns:")
print(test.columns.tolist())

print("\nFirst train row:")
print(train.iloc[0].to_dict())

print("\nFirst test row:")
print(test.iloc[0].to_dict())
```

```python
# ============================================================
# DRUG IDENTITY OVERLAP — PAIR-DISJOINT TRAIN vs TEST
# ============================================================

train_drugs = set(train["Drug1"].astype(str)) | set(train["Drug2"].astype(str))
test_drugs  = set(test["Drug1"].astype(str))  | set(test["Drug2"].astype(str))

overlap_drugs = train_drugs & test_drugs

train_only = train_drugs - test_drugs
test_only  = test_drugs - train_drugs

print("=" * 70)
print("DRUG IDENTITY OVERLAP — PAIR-DISJOINT SPLIT")
print("=" * 70)

print(f"\nUnique drugs in Train : {len(train_drugs)}")
print(f"Unique drugs in Test  : {len(test_drugs)}")
print(f"Shared drugs          : {len(overlap_drugs)}")

print(f"\nTrain-only drugs      : {len(train_only)}")
print(f"Test-only drugs       : {len(test_only)}")

print("\nOVERLAP RATES")
print("-" * 70)

print(
    f"Test drugs also present in Train : "
    f"{len(overlap_drugs) / len(test_drugs) * 100:.2f}%"
)

print(
    f"Train drugs also present in Test : "
    f"{len(overlap_drugs) / len(train_drugs) * 100:.2f}%"
)

# ------------------------------------------------------------
# Pair overlap verification
# ------------------------------------------------------------

train_pairs = set(train["Pair_ID"].astype(str))
test_pairs  = set(test["Pair_ID"].astype(str))

pair_overlap = train_pairs & test_pairs

print("\nPAIR OVERLAP CHECK")
print("-" * 70)

print(f"Unique Train pairs : {len(train_pairs)}")
print(f"Unique Test pairs  : {len(test_pairs)}")
print(f"Shared pairs       : {len(pair_overlap)}")

if len(pair_overlap) == 0:
    print("\n✓ Pair-disjoint condition confirmed.")
else:
    print("\nWARNING: Pair overlap detected!")

print("\n✓ Drug-overlap audit complete.")
```

```python
from pathlib import Path
import pandas as pd

ROOT = Path(r"C:\Users\user\Desktop\drug drug\drug2")

print("=" * 80)
print("SEARCHING FOR APIXABAN / CLOPIDOGREL IN PROJECT CSV FILES")
print("=" * 80)

terms = [
    "apixaban",
    "clopidogrel",
    "DB06605",
    "DB00758"
]

hits = []

for f in ROOT.rglob("*.csv"):
    try:
        df = pd.read_csv(f, dtype=str, low_memory=False)

        # Search all columns without printing huge tables
        mask = pd.Series(False, index=df.index)

        for col in df.columns:
            s = df[col].astype(str).str.lower()

            for term in terms:
                mask |= s.str.contains(term.lower(), regex=False, na=False)

        n = int(mask.sum())

        if n > 0:
            hits.append((str(f), n))

            print("\nFILE:")
            print(f)
            print("Matching rows:", n)

            # Print only a few matching rows
            print(df.loc[mask].head(10).to_string(index=False))

    except Exception:
        pass

print("\n" + "=" * 80)
print("SEARCH COMPLETE")
print("=" * 80)

print("Files containing matches:", len(hits))

for f, n in hits:
    print(n, "rows ->", f)
```

```python
from pathlib import Path
import pandas as pd

ROOT = Path(r"C:\Users\user\Desktop\drug drug\drug2")

A = "DB06605"   # Apixaban
B = "DB00758"   # Clopidogrel

print("=" * 80)
print("EXACT PAIR SEARCH: APIXABAN + CLOPIDOGREL")
print("=" * 80)

found = []

for f in ROOT.rglob("*.csv"):
    try:
        df = pd.read_csv(f, dtype=str, low_memory=False)

        # Search each row for BOTH IDs
        text = df.fillna("").astype(str).agg(" ".join, axis=1)

        mask = (
            text.str.contains(A, regex=False) &
            text.str.contains(B, regex=False)
        )

        if mask.any():
            sub = df.loc[mask]

            print("\nFILE:")
            print(f)
            print("Exact-pair rows:", len(sub))
            print("Columns:", df.columns.tolist())

            # only relevant columns where possible
            useful = [
                c for c in df.columns
                if any(k in c.lower() for k in [
                    "drug", "pair", "interaction",
                    "level", "pk", "pd",
                    "target", "predict"
                ])
            ]

            if useful:
                print(sub[useful].head(20).to_string(index=False))
            else:
                print(sub.head(20).to_string(index=False))

            found.append((str(f), len(sub)))

    except Exception:
        pass

print("\n" + "=" * 80)

if found:
    print("EXACT PAIR FOUND")
    print("Files:", len(found))
    for f, n in found:
        print(n, "row(s) ->", f)
else:
    print("EXACT PAIR NOT FOUND IN CSV FILES")

print("=" * 80)
```

```python
from pathlib import Path
import pandas as pd

BASE = Path(r"C:\Users\user\Desktop\drug drug\drug2\results_level2_pair_disjoint")

A = "DB00758"
B = "DB06605"

print("="*80)
print("APIXABAN–CLOPIDOGREL: REPRESENTATIVE / FIGURE-5 RESULT")
print("="*80)

keywords = [
    "representative",
    "figure",
    "case",
    "prediction",
    "hierarchical"
]

found = []

for f in BASE.rglob("*.csv"):
    name = f.name.lower()

    if not any(k in name for k in keywords):
        continue

    try:
        df = pd.read_csv(f, dtype=str, low_memory=False)

        text = df.fillna("").astype(str).agg(" ".join, axis=1)

        mask = (
            text.str.contains(A, regex=False) &
            text.str.contains(B, regex=False)
        )

        if mask.any():

            sub = df.loc[mask]

            print("\nFILE:")
            print(f)
            print("Rows:", len(sub))
            print("Columns:", df.columns.tolist())
            print()
            print(sub.to_string(index=False))

            found.append(str(f))

    except Exception:
        pass

print("\n" + "="*80)
print("Files containing the exact pair:", len(found))
for f in found:
    print(f)
```
